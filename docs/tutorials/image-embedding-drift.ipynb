# Image Embedding Drift Tutorial

This notebook shows image drift checks with a lightweight custom embedder and the high-level helper APIs.

````python
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().resolve().parents[1]))
import numpy as np
from drift_control.image_embedding_drift import detect_image_embedding_drift
````

````python
rng = np.random.default_rng(5)
reference_images = [rng.normal(0.0, 1.0, size=(16, 16, 3)).astype(np.float32) for _ in range(200)]
current_images = [rng.normal(0.6, 1.0, size=(16, 16, 3)).astype(np.float32) for _ in range(200)]
len(reference_images), reference_images[0].shape
````

````python
def simple_embedder(images):
    arr = np.asarray(images, dtype=float)
    flat = arr.reshape(arr.shape[0], -1)
    return flat[:, :64]  # simple projection for tutorial/demo

result = detect_image_embedding_drift(
    reference_images=reference_images,
    current_images=current_images,
    method='energy',
    embedder=simple_embedder,
    alpha=0.05,
    n_permutations=80,
    random_state=7,
)
result
````
