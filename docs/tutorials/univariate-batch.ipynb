# Univariate Batch Drift Tutorial

This notebook shows a minimal batch workflow for per-column drift checks using `drift-control`.

````python
import numpy as np
import pandas as pd

from drift_control import UnifiedDriftDetector
````

````python
rng = np.random.default_rng(42)
baseline_df = pd.DataFrame({
    'amount': rng.normal(100, 20, size=600),
    'latency_ms': rng.normal(250, 40, size=600),
})
current_df = pd.DataFrame({
    'amount': rng.normal(118, 24, size=600),
    'latency_ms': rng.normal(252, 40, size=600),
})
baseline_df.head()
````

````python
ks = UnifiedDriftDetector(method='ks', alpha=0.05)
psi = UnifiedDriftDetector(method='psi', threshold=0.2)

rows = []
for col in baseline_df.columns:
    ks_res = ks.detect_drift(baseline_df[col], current_df[col])
    psi_res = psi.detect_drift(baseline_df[col], current_df[col])
    rows.append({
        'column': col,
        'ks_drift': ks_res.drift,
        'ks_p_value': ks_res.p_value,
        'psi_drift': psi_res.drift,
        'psi_score': psi_res.score,
    })

pd.DataFrame(rows)
````
