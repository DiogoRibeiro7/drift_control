# Embedding Drift Tutorial (Precomputed Vectors)

This notebook demonstrates drift checks on precomputed embeddings using multivariate detectors.

```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().resolve().parents[1]))
import numpy as np
from drift_control.unified_drift_detector import UnifiedDriftDetector
```

```python
rng = np.random.default_rng(11)
reference_embeddings = rng.normal(0.0, 1.0, size=(800, 32))
current_embeddings = rng.normal(0.45, 1.0, size=(800, 32))
reference_embeddings.shape, current_embeddings.shape
```

```python
mmd = UnifiedDriftDetector(method='mmd', alpha=0.05, n_permutations=120, random_state=7)
mmd_result = mmd.detect_drift(reference_embeddings, current_embeddings)
mmd_result
```

```python
energy = UnifiedDriftDetector(method='energy', alpha=0.05, n_permutations=120, random_state=7)
energy_result = energy.detect_drift(reference_embeddings, current_embeddings)
energy_result
```

## Notes

- Use precomputed embeddings from your text/image pipeline to avoid coupling detector code to model inference.
- If compute budget is tight, start with `energy`; if sensitivity is the priority, compare with `mmd`.
